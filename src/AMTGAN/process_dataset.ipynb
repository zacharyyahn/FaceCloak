```python
import os
import shutil
from tqdm import tqdm

old_path = "assets/datasets/CelebA-HQ/"
new_path = "../../data/celeba_small/probe/"

count = 0
for im in tqdm(os.listdir(old_path)):
    count += 1
    name = im[:im.find(".")]
    shutil.copy(old_path+im,new_path+name+"_1.jpg")
    if count > 100:
        break
```
